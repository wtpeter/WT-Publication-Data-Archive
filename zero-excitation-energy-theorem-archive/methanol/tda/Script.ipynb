```python
import numpy as np
import pandas as pd
import os
```

```python
fun_list = ['SVWN', 'BLYP', 'PBE', 'TPSS', 'B3LYP', 'PBE0']
data = {}
for fun in fun_list:
    temp = {}
    temp['SC'] = np.load(f'sc_{fun}.npy')[0]
    temp['SF'] = (np.load(f'sfu_{fun}.npy')[0] - np.load(f'sc_{fun}.npy')[0])
    data[fun] = temp
```

```python
df = pd.DataFrame(data)
```

```python
df
```

```python
# Format all values in the dataframe to 4 decimal places as strings
df_formatted = df.applymap(lambda x: f"{x:.4f}")

# Display the formatted dataframe
df_formatted
```
